#### Task 3 : Working with Running Times
Instructions : Create two variable lists and assign them numeric values and perform the following operations :
##### A. Add Both runs to get the total time spent running.
##### B. Subtract run2 from run1 to see the improvement.
##### C. Multiple run2 by 2 to simulate doing it twice.
##### D. Calculate the average of both runs.

```python
run_time_one=40
run_time_two=15
total_time = run_time_one+run_time_two
improvement= run_time_one-run_time_two
double_run=run_time_two*2
average=(run_time_one+run_time_two)/2

print("The total running time was",total_time,"minutes")
print("Improvement:", improvement, "minutes")
print("Run 2 done twice:", double_run, "minutes")
print("Average running time:", average, "minutes")
```

 #### Task 4 : String Manipulation

 #### Task 5 : User Input and Operations.
Instructions : Recalling the Pythagorean theorem complete the following operations : 
##### A. Prompt a user to enter the lengths of the two legs. 
##### B. Use the formula for the Pythagorean theorem to compute the hypotenuse.
##### C. Print the following statement: “The hypotenuse of your triangle is <c> units.” Where <c> is the computed hypotenuse, rounded to the nearest one-hundredth.

```python
leg1 = float(input("Enter the length of the first leg: "))
leg2 = float(input("Enter the length of the second leg: "))

hypotenuse = (leg1 ** 2 + leg2 ** 2) ** 0.5

print("The hypotenuse of your triangle is", round(hypotenuse, 2), "units.")
```
