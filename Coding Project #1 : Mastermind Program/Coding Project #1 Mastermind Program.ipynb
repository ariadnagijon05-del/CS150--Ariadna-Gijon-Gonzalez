```python

import random

# Generate a secret four-digit number
def generate_secret():
    secret = ""

    for i in range(4):
        digit = random.randint(0, 9)
        secret = secret + str(digit)

    return secret


# Compare the player's guess with the secret number
def check_guess(secret, guess):

    correct_position = 0
    wrong_position = 0

    secret_list = list(secret)
    guess_list = list(guess)

    # Find digits in the correct position
    for i in range(4):
        if secret_list[i] == guess_list[i]:
            correct_position = correct_position + 1

    # Find matching digits in the wrong position
    secret_remaining = []
    guess_remaining = []

    for i in range(4):
        if secret_list[i] != guess_list[i]:
            secret_remaining.append(secret_list[i])
            guess_remaining.append(guess_list[i])

    for digit in guess_remaining:
        if digit in secret_remaining:
            wrong_position = wrong_position + 1
            secret_remaining.remove(digit)

    return correct_position, wrong_position


# Play one game
def play_game():

    secret = generate_secret()
    attempts = 10

    print("\nWelcome to Mastermind!")
    print("Guess the secret four-digit number.")
    print("You have 10 attempts.")

    for attempt in range(1, attempts + 1):

        guess = input("\nEnter your four-digit guess: ")

        # Validate the guess
        while len(guess) != 4 or not guess.isdigit():
            print("Invalid input. Please enter exactly four digits.")
            guess = input("Enter your four-digit guess: ")

        correct, wrong = check_guess(secret, guess)

        print("Correct position:", correct)
        print("Correct digit, wrong position:", wrong)

        if correct == 4:
            print("Congratulations! You won!")
            return True

        print("Attempts remaining:", attempts - attempt)

    print("\nYou lost!")
    print("The secret number was:", secret)

    return False


# Keep track of wins and losses
def main():

    wins = 0
    losses = 0

    play_again = "yes"

    print("Welcome to the Mastermind Game!")

    while play_again == "yes":

        result = play_game()

        if result == True:
            wins = wins + 1
        else:
            losses = losses + 1

        print("\nGame Record")
        print("Wins:", wins)
        print("Losses:", losses)

        play_again = input("\nWould you like to play again? (yes/no): ")
        play_again = play_again.lower()

    print("\nThanks for playing!")


# Start the game
main()
```
