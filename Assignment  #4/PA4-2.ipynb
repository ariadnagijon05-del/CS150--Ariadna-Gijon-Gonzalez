## Task 1: Input Positive Integer and Factors

This task asks the user for a positive integer. The program displays all of its positive factors and determines whether the number is prime.

```python
number = int(input("Please input a positive integer: "))

for factor in range(1, number + 1):
    if number % factor == 0:
        print(factor, "is a factor of", number)
```

```python
number = int(input("Enter a positive integer: "))

count = 0

for i in range(1, number + 1):
    if number % i == 0:
        count = count + 1

if count == 2:
    print("The number is prime")
else:
    print("The number is not prime")
```

## Task 2: Language Translator

This program uses a dictionary to translate 10 English words into Spanish. It continues asking the user for words until they press Enter without typing anything.

```python
translation_for_a_dictionary = {
    "hello": "hola",
    "goodbye": "adios",
    "please": "por favor",
    "thank you": "gracias",
    "water": "agua",
    "food": "comida",
    "house": "casa",
    "book": "libro",
    "dog": "perro",
    "cat": "gato"
}
```

```python
word = input("Enter an English word to translate (press Enter to stop): ")

while word != "":
    word = word.lower()

    if word in translation_for_a_dictionary:
        print("The Spanish translation is", translation_for_a_dictionary[word])
    else:
        print("That word is not in the dictionary.")

    word = input("Enter an English word to translate (press Enter to stop): ")
```

## Task 3: Temperature Summaries

This program stores the average high and low temperatures for each month in Louisville, Kentucky. The user can enter a month to see its average temperatures or enter quit to stop the program.

```python
temps = {
    "January": [40, 20],
    "February": [42, 31],
    "March": [56, 38],
    "April": [60, 49],
    "May": [70, 59],
    "June": [86, 67],
    "July": [89, 81],
    "August": [88, 80],
    "September": [82, 52],
    "October": [71, 30],
    "November": [58, 29],
    "December": [47, 32]
}
```

```python
month = input("Enter a month (or quit to stop): ")

while month.lower() != "quit":

    month = month.capitalize()

    if month in temps:
        print("Average high:", temps[month][0])
        print("Average low:", temps[month][1])
    else:
        print("Not a valid month.")

    month = input("Enter a month (or quit to stop): ")
```

## Task 4: Inventory Dictionary and Prices

This program stores items, quantities, and prices in an inventory dictionary. The user can look up an item, find items below a budget, calculate the total quantity, or calculate the total value of the inventory.

```python
inventory = {
    'hammer': [15, 12.49],
    'wrench': [20, 7.99],
    'screwdriver': [40, 3.99],
    'nailgun': [10, 36.59],
    'padlock': [25, 5.99],
    'roll_of_duct_tape': [30, 4.95]
}
```

```python
print("Inventory Menu")
print("1. Look up an item")
print("2. Find what is below a budget")
print("3. Find the total quantity")
print("4. Find the total inventory value")

choice = input("Enter a menu option (1-4): ")
```

```python
if choice == "1":
    item = input("Enter an item name: ")

    if item in inventory:
        print("Quantity available:", inventory[item][0])
        print("Price per item: $", inventory[item][1])
    else:
        print("That item is not available right now.")

elif choice == "2":
    budget = float(input("Enter your max budget: $"))

    for item in inventory:
        if inventory[item][1] < budget:
            print(item, "costs $", inventory[item][1])

elif choice == "3":
    total_quantity = 0

    for item in inventory:
        total_quantity = total_quantity + inventory[item][0]

    print("The total quantity of all items is", total_quantity)

elif choice == "4":
    total_price = 0

    for item in inventory:
        total_price = total_price + inventory[item][0] * inventory[item][1]

    print("The total value of all inventory is $", round(total_price, 2))

else:
    print("Invalid : option does not exist.")
```
