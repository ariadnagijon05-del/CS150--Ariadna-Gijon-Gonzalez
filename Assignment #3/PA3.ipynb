#### Task 1 : Types of Triangles
Instructions : For this task, create a simple program that does the following...
##### A. Takes user input representing three sides of a triangle.
##### B. Determines weather the three sides represent an equilateral triangle, an isoscekes (but not equilateral) triangle or a scalene triangle. The output should be given to the user.

```python
side1 = float(input("Enter the first side: "))
side2 = float(input("Enter the second side: "))
side3 = float(input("Enter the third side: "))

if side1==side2 and side2==side3:
    print("The triangle is equilateral")
elif side1==side2 or side1==side3 or side2==side3:
    print("The triangle is isosceles")
else:
    print("The triangle is scalene")
```

#### Task 2 : Student Grades and At-Risk Status
Instructions : 
##### A. Read five input grade averages (which should be floats) representing student grades at the mid-term break into an initially empty list called student_grades.
##### B. Calculate and print the class average, minimum grade, and maximum grade.
##### C. Any student with a grade below 65 is considered “at risk” – as such, you should print a message indicating how many of the students in the course are at risk. 

```python
grade1=float(input("Enter the grade"))
grade2=float(input("Enter the grade"))
grade3=float(input("Enter the grade"))
grade4=float(input("Enter the grade"))
grade5=float(input("Enter the grade"))

student_grades=[grade1,grade2,grade3,grade4,grade5]

print(sum(student_grades)/len(student_grades),"is the average grade") #average
print(min(student_grades),"is the minimum grade") #minimum
print(max(student_grades),"is the maximum grade") #maximum

atriskcounter=0

for grade in student_grades:
    if grade<65:
        atriskcounter+=1

print("The number of students at risk are ",atriskcounter)
```

#### Task 3 : Ranges and Multiples
Instructions : 
##### A. Create a list called mult_of_2 that contains all multiples of 2 between 1 and 100.
##### B. Create a list called mult_of_3 that contains all multiples of 3 between 1 and 100.
##### C. Create a list called mult_of_both that contains all elements that are present in both mult_of_2 and mult_of_3.

```python
mult_of_2 = 2
while mult_of_2<=100:
    print(mult_of_2)
    mult_of_2=(mult_of_2)+2
print("Done. These are the multiples of 2 between 1 and 100")
```

```python
mult_of_3 = 3
while mult_of_3<=100:
    print(mult_of_3)
    mult_of_3=(mult_of_3)+3
print("Done. These are the multiples of 3 between 1 and 100")
```

```python
mult_of_both = []

mycount = 1
while mycount<=100:
    if mycount%2==0 and mycount%3==0:
        mult_of_both.append(mycount)
    mycount=mycount+1

print(mult_of_both)
```

#### Task 4 : Shopping List
Instructions : 
##### A. Starting with an initially empty list named shopping_list, allow the user to input up to 7 items they want to purchase (an input value like “done” should be used to stop input early.)
##### B. Print the total number of items on your shopping list. 
##### C. Print the list in reverse order (opposite order from which the values were input).
##### D. Print the list in alphabetical order.
##### E. If any item appears at least twice in the list, let the person know as much.  Hint: you may want to use embedded loops for this task.

```python
shopping_list = []

mycount = 1

while mycount<=7:
    item = input("Enter an item or type done: ")

    if item=="done":
        break

    shopping_list.append(item)
    mycount=mycount+1

print("You have",len(shopping_list),"items on your shopping list.")

print("Your shopping list is:")
print(shopping_list)

shopping_list.reverse()
print("Your list in reverse is:")
print(shopping_list)

shopping_list.sort()
print("Your list in alphabetical order is:")
print(shopping_list)

for item in shopping_list:
    if shopping_list.count(item)>1:
        print(item,"is a duplicate")
```
