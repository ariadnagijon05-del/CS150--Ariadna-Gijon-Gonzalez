#### Task #1 : Weekend Activites as Sets 

```python
activities_saturday = set()

activity = input('Enter an activity from last Saturday ("Done" to stop): ')

while activity != "Done":
    activities_saturday.add(activity)

    activity = input('Enter an activity from last Saturday ("Done" to stop): ')
```

```python
activites_sunday = set()

activity = input('Enter an activity from last Sunday ("Done" to stop): ')

while activity != "Done":
    activites_sunday.add(activity)

    activity = input('Enter an activity from last Sunday ("Done" to stop): ')
```

```python
print("Saturday activities:", activities_saturday)
print("Sunday activities:", activites_sunday)
```

```python
print("Activities done on both days:")

for activity in activities_saturday:
    if activity in activites_sunday:
        print(activity)
```

```python
print("Activities done on at least one day:")

for activity in activities_saturday:
    print(activity)

for activity in activites_sunday:
    if activity not in activities_saturday:
        print(activity)
```

#### Task #2 : Triangle Calculations

```python
def triangle_solve(leg1, leg2):

    hypotenuse = (leg1**2 + leg2**2)**0.5

    perimeter = leg1 + leg2 + hypotenuse

    area = (leg1 * leg2) / 2

    return (hypotenuse, perimeter, area)
```

```python
triangle = triangle_solve(6, 8)

print(triangle)
```

#### Task #3: Calculator Class

```python
class Calculator:

    def __init__(self):
        self.previous_result = 0
        self.total_calculations = 0

    def add(self, a, b):
        result = a + b
        self.previous_result = result
        self.total_calculations += 1
        return result

    def subtract(self, a, b):
        result = a - b
        self.previous_result = result
        self.total_calculations += 1
        return result

    def multiply(self, a, b):
        result = a * b
        self.previous_result = result
        self.total_calculations += 1
        return result

    def divide(self, a, b):
        result = a / b
        self.previous_result = result
        self.total_calculations += 1
        return result

    def power(self, a, b):
        result = a ** b
        self.previous_result = result
        self.total_calculations += 1
        return result

    def info(self):
        print("Previous result:", self.previous_result)
        print("Total calculations:", self.total_calculations)
```

```python
calc = Calculator()

print(calc.add(5, 3))
print(calc.subtract(10, 4))
print(calc.multiply(2, 5))
print(calc.divide(20, 4))
print(calc.power(2, 3))

calc.info()
```
